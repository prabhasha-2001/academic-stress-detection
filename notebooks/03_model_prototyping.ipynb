```python
import sys, pathlib
sys.path.append(str(pathlib.Path.cwd().parent))
import matplotlib.pyplot as plt
from src.config import *
```

```python
from src.evaluation.cross_validate import load_dataset, make_split
from src.models.train_logistic_regression import get_model as lr
from src.models.train_decision_tree import get_model as dt
from src.models.train_random_forest import get_model as rf
X, y = load_dataset('multiclass')
X_tr, X_te, y_tr, y_te = make_split(X, y)
print(X_tr.shape, X_te.shape)
```

```python
from sklearn.model_selection import cross_val_score
from src.evaluation.cross_validate import cv_splitter
for name, get in [('LR', lr), ('DT', dt), ('RF', rf)]:
    pipe, _ = get()
    s = cross_val_score(pipe, X_tr, y_tr, cv=cv_splitter(), scoring='f1_macro')
    print(name, 'macro-F1 %.3f +/- %.3f' % (s.mean(), s.std()))
```
