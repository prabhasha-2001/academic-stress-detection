```python
import sys, pathlib
sys.path.append(str(pathlib.Path.cwd().parent))
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
from src.config import *
```

```python
df = pd.read_csv(DATA_LABELED)
df[['age','pss_total']].describe()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(9, 3))
df[TARGET].value_counts().reindex(CLASS_ORDER).plot.bar(ax=ax[0], title='Class distribution')
df['pss_total'].plot.hist(bins=20, ax=ax[1], title='PSS-10 total')
plt.tight_layout(); plt.savefig(FIGURES/'eda_label_distribution.png', dpi=200)
```

```python
from scipy.stats import spearmanr
from src.preprocessing.encode_features import encode_ordinals
enc = encode_ordinals(df)
rho = {c: spearmanr(enc[c], enc['pss_total'])[0] for c in NUMERIC_FEATURES}
pd.Series(rho).sort_values(key=abs, ascending=False).round(3)
```
